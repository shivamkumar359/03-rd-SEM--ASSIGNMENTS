```python
# ==========================================================
# TASK 1: IMPLEMENT K-FOLD CROSS-VALIDATION
# ==========================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.backends.backend_pdf import PdfPages
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
```

```python
# ==========================================================
# STEP 1: LOAD SAMPLE DATASET
# ==========================================================

iris = load_iris()

X = iris.data
y = iris.target

sample_ids = np.arange(len(y))

print("Dataset loaded successfully!")
print("Total samples:", len(y))
print("Number of classes:", len(iris.target_names))
```

```python
# ==========================================================
# STEP 2: RANDOMLY SHUFFLE DATASET
# ==========================================================

random_seed = 42

rng = np.random.default_rng(random_seed)

shuffled_indices = rng.permutation(len(y))

X = X[shuffled_indices]
y = y[shuffled_indices]
sample_ids = sample_ids[shuffled_indices]

print("\nDataset shuffled successfully!")
```

```python
# ==========================================================
# STEP 3: CREATE FIVE EQUAL FOLDS
# ==========================================================

k = 5

folds = np.array_split(np.arange(len(y)), k)

print("\nDataset divided into five folds:")

for i, fold in enumerate(folds):
    print(f"Fold {i + 1}: {len(fold)} samples")
```
