```python
# Bayes Theorem Calculator

prior = 0.50
likelihood = 0.80
prob_not_A = 1 - prior
prob_B_given_not_A = 0.50

# Calculate total probability of evidence
prob_B = (likelihood * prior) + (prob_B_given_not_A * prob_not_A)

if prob_B > 0:
    posterior = (likelihood * prior) / prob_B

    print("Prior Probability:", prior)
    print("Likelihood:", likelihood)
    print("Marginal Likelihood P(B):", prob_B)
    print("Posterior Probability:", posterior)
    print("Posterior Percentage:", posterior * 100, "%")
else:
    print("Error: P(B) must be greater than 0.")
```
