```python
# TASK 2: Evaluate Bagging and Out-of-Bag Error

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import BaggingRegressor
from sklearn.metrics import r2_score, mean_squared_error

```

```python
# --------------------------------------------------
# 1. Create the Housing Dataset
# --------------------------------------------------

data = {
    "House_ID": [1, 2, 3, 4, 5, 6],
    "Length": [15, 20, 12, 25, 18, 22],
    "Width": [10, 15, 8, 20, 12, 18],
    "Rooms": [3, 4, 2, 5, 3, 4],
    "Age": [5, 2, 20, 1, 10, 8],
    "Price": [250000, 450000, 180000, 700000, 310000, 520000]
}

df = pd.DataFrame(data)
```

```python
# --------------------------------------------------
# 2. Separate Features and Target
# --------------------------------------------------

X = df[["Length", "Width", "Rooms", "Age"]]
y = df["Price"]
```

```python
# --------------------------------------------------
# 3. Split Dataset into Training and Testing Sets
# --------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.33,
    random_state=42
)

```

```python
# --------------------------------------------------
# 4. Train a Single Decision Tree
# --------------------------------------------------

single_tree = DecisionTreeRegressor(random_state=42)

single_tree.fit(X_train, y_train)

single_predictions = single_tree.predict(X_test)

single_r2 = r2_score(y_test, single_predictions)
single_mse = mean_squared_error(y_test, single_predictions)

# --------------------------------------------------
# 5. Configure the Bagging Regressor
# --------------------------------------------------

bagging_model = BaggingRegressor(
    estimator=DecisionTreeRegressor(random_state=42),
    n_estimators=50,
    max_samples=1.0,
    bootstrap=True,
    oob_score=True,
    n_jobs=-1,
    random_state=42
)

# --------------------------------------------------
# 6. Train the Bagging Model
# --------------------------------------------------

bagging_model.fit(X_train, y_train)

# --------------------------------------------------
# 7. Generate Predictions
# --------------------------------------------------

bagging_predictions = bagging_model.predict(X_test)

# --------------------------------------------------
# 8. Calculate OOB Score and Test Performance
# --------------------------------------------------

oob_score = bagging_model.oob_score_

bagging_r2 = r2_score(y_test, bagging_predictions)

bagging_mse = mean_squared_error(y_test, bagging_predictions)

# --------------------------------------------------
# 9. Display Results
# --------------------------------------------------

print("\n========== BAGGING MODEL EVALUATION ==========")

print("Number of Base Models:", bagging_model.n_estimators)
print("Base Estimator: DecisionTreeRegressor")
print("OOB Scoring:", bagging_model.oob_score)
print("Parallel Processing:", bagging_model.n_jobs)

print("\n--- Single Decision Tree ---")
print("Test R-squared:", round(single_r2, 4))
print("Test MSE:", round(single_mse, 2))

print("\n--- Bagging Regressor ---")
print("OOB R-squared:", round(oob_score, 4))
print("Test R-squared:", round(bagging_r2, 4))
print("Test MSE:", round(bagging_mse, 2))

print("==============================================")

# --------------------------------------------------
# 10. Save Parameter Log
# --------------------------------------------------

with open("parameter_log.txt", "w", encoding="utf-8") as file:
    file.write("BAGGING MODEL PARAMETER LOG\n")
    file.write("===========================\n\n")
    file.write("Base Estimator: DecisionTreeRegressor\n")
    file.write(f"Number of Base Models: {bagging_model.n_estimators}\n")
    file.write(f"OOB Scoring: {bagging_model.oob_score}\n")
    file.write(f"Bootstrap Sampling: {bagging_model.bootstrap}\n")
    file.write(f"Parallel Processing (n_jobs): {bagging_model.n_jobs}\n")
    file.write(f"Random State: {bagging_model.random_state}\n")

print("\nParameter log saved to parameter_log.txt")
```
