```python
# Task 1: Build a Random Forest Regressor
# File: rf_housing.py

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score, mean_squared_error
```

```python
# --------------------------------------------------
# 1. Create the Housing Dataset
# --------------------------------------------------

data = {
    "House_ID": [1, 2, 3, 4, 5, 6],
    "Length": [15, 20, 12, 25, 18, 22],
    "Width": [10, 15, 8, 20, 12, 18],
    "Rooms": [3, 4, 2, 5, 3, 4],
    "Age": [5, 2, 20, 1, 10, 8],
    "Price": [250000, 450000, 180000, 700000, 310000, 520000]
}

df = pd.DataFrame(data)

```

```python
# --------------------------------------------------
# 2. Separate Features and Target Variable
# --------------------------------------------------

# House_ID is excluded because it is only an identifier.
X = df[["Length", "Width", "Rooms", "Age"]]
y = df["Price"]

# --------------------------------------------------
# 3. Split the Dataset
# --------------------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.33,
    random_state=42
)

# --------------------------------------------------
# 4. Initialize the Random Forest Regressor
# --------------------------------------------------

model = RandomForestRegressor(
    n_estimators=100,
    max_features=2,
    bootstrap=True,
    random_state=42
)

# --------------------------------------------------
# 5. Train the Model
# --------------------------------------------------

model.fit(X_train, y_train)

# --------------------------------------------------
# 6. Predict House Prices
# --------------------------------------------------

y_pred = model.predict(X_test)

# --------------------------------------------------
# 7. Evaluate the Model
# --------------------------------------------------

r2 = r2_score(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)

print("\n========== MODEL EVALUATION ==========")
print("Model: Random Forest Regressor")
print("Number of Trees:", model.n_estimators)
print("Maximum Features per Split:", model.max_features)
print("Bootstrap Sampling:", model.bootstrap)
print("Training Samples:", len(X_train))
print("Testing Samples:", len(X_test))
print("R-squared (R2):", round(r2, 4))
print("Mean Squared Error (MSE):", round(mse, 2))
print("======================================")

# --------------------------------------------------
# 8. Display Feature Importance
# --------------------------------------------------

importance_df = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
)

print("\n========== FEATURE IMPORTANCE ==========")
print(importance_df.to_string(index=False))
print("========================================")

# --------------------------------------------------
# 9. Save Predictions to CSV
# --------------------------------------------------

predictions_df = pd.DataFrame({
    "House_ID": df.loc[X_test.index, "House_ID"].values,
    "Actual_Price": y_test.values,
    "Predicted_Price": y_pred
})

predictions_df.to_csv("predictions.csv", index=False)

print("\nPredictions saved to predictions.csv")

# --------------------------------------------------
# 10. Save Feature Importance to Text File
# --------------------------------------------------

with open("feature_importance.txt", "w", encoding="utf-8") as file:
    file.write("RANDOM FOREST FEATURE IMPORTANCE\n")
    file.write("================================\n\n")

    for rank, row in enumerate(
        importance_df.itertuples(index=False),
        start=1
    ):
        file.write(
            f"{rank}. {row.Feature}: {row.Importance:.4f}\n"
        )

print("Feature importance saved to feature_importance.txt")
```
