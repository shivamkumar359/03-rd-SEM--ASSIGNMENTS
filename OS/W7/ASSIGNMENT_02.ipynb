```python
# ============================================================
# Data Filtering and Statistical Summarization
# ============================================================

# Step 1: Import the required library
import pandas as pd
```

```python
# Step 2: Create a sample employee dataset
# Replace this section with your actual dataset when available.

data = {
    "EmployeeID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Name": [
        "Aman", "Rahul", "Priya", "Neha", "Arjun",
        "Sneha", "Vikash", "Riya", "Karan", "Anjali"
    ],
    "Salary": [
        35000, 45000, 55000, 38000, 65000,
        42000, 30000, 75000, 50000, 40000
    ],
    "YearsExperience": [1, 3, 5, 2, 7, 4, 1, 9, 6, 3],
    "Age": [23, 25, 29, 24, 32, 27, 22, 35, 30, 26]
}

df = pd.DataFrame(data)
```

```python
# Step 3: Display the original dataset

print("=" * 65)
print("DATA FILTERING AND STATISTICAL SUMMARIZATION")
print("=" * 65)

print("\nOriginal Dataset:")
print(df)

```

```python
#  Step 4: Display dataset information using info()

print("\nDataset Information")
print("-" * 65)

df.info()
```

```python
# Step 5: Generate descriptive statistics using describe()

print("\nStatistical Summary")
print("-" * 65)

print(df.describe())
```

```python
# Step 6: Select rows using iloc()

print("\nSelecting Rows Using iloc()")
print("-" * 65)

# Display the first five rows using integer positions
print(df.iloc[0:5])
```

```python
# Step 7: Filter rows using loc()
# Select employees whose salary exceeds 40,000.

print("\nFiltering Employees with Salary > 40000")
print("-" * 65)

filtered_df = df.loc[df["Salary"] > 40000].copy()

print(filtered_df)
```

```python
# Step 8: Apply clipping to cap numerical values

print("\nApplying Clip Function")
print("-" * 65)

# Cap salaries between 30,000 and 70,000.
# This creates a separate clipped version for demonstration.

clipped_df = df.copy()

clipped_df["Salary"] = clipped_df["Salary"].clip(
    lower=30000,
    upper=70000
)

print(clipped_df[["EmployeeID", "Name", "Salary"]])
```

```python
#  Step 9: Calculate mean years of experience
# Use the filtered subset where salary exceeds 40,000.

print("\nMean Years of Experience")
print("-" * 65)

if not filtered_df.empty:
    mean_experience = filtered_df["YearsExperience"].mean()

    print(
        "Mean Years of Experience for employees with "
        f"Salary > 40000: {mean_experience:.2f} years"
    )

    # Save the calculated mean to a text file.
    with open("mean_calc.txt", "w", encoding="utf-8") as file:
        file.write(
            "Mean Years of Experience for Employees "
            "with Salary > 40000: "
            f"{mean_experience:.2f} years\n"
        )

    print("Mean calculation saved to mean_calc.txt")

else:
    print("No employees satisfy the salary condition.")
```

```python
# Step 10: Export the filtered dataset to CSV

filtered_df.to_csv(
    "subset_results.csv",
    index=False
)

print("\nFiltered dataset saved to subset_results.csv")
```

```python
# Step 11: Display statistical information for the filtered subset

print("\nFiltered Dataset Information")
print("-" * 65)

filtered_df.info()

print("\nFiltered Dataset Statistical Summary")
print("-" * 65)

print(filtered_df.describe())
```
