```python
# ============================================================
# Task: Implement and Evaluate a Random Forest Regressor
# ============================================================

# Step 1: Import required libraries

import pandas as pd

from sklearn.datasets import make_regression
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)
```

```python
# Step 2: Generate a sample regression dataset
# Replace this section with your provided dataset when available.

X, y = make_regression(
    n_samples=1000,
    n_features=8,
    n_informative=6,
    noise=15,
    random_state=42
)

feature_names = [f"Feature_{i+1}" for i in range(X.shape[1])]

X = pd.DataFrame(X, columns=feature_names)
y = pd.Series(y, name="Target")
```

```python
# Step 3: Display dataset information

print("=" * 60)
print("RANDOM FOREST REGRESSOR")
print("=" * 60)

print("\nDataset Information")
print("-" * 60)

print("Number of samples:", X.shape[0])
print("Number of input features:", X.shape[1])
print("Target variable:", y.name)
```

```python
# Step 4: Split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nData Splitting")
print("-" * 60)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
# Step 5: Initialize the Random Forest Regressor

model = RandomForestRegressor(
    n_estimators=100,
    criterion="squared_error",
    max_features="sqrt",
    bootstrap=True,
    random_state=42,
    n_jobs=-1
)
```

```python
# Step 6: Train the Random Forest model

print("\nTraining Model")
print("-" * 60)

model.fit(X_train, y_train)

print("Random Forest model trained successfully.")
```

```python
# Step 7: Generate predictions

y_pred = model.predict(X_test)

print("Predictions generated successfully.")
```

```python
# Step 8: Evaluate model performance

mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("\nModel Evaluation Results")
print("-" * 60)

print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"R-squared (R2 Score): {r2:.4f}")
```

```python
# Step 9: Extract feature importance scores

importance_df = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": model.feature_importances_
})
```

```python
# Step 10: Rank features by importance

importance_df = importance_df.sort_values(
    by="Importance",
    ascending=False
).reset_index(drop=True)

importance_df.insert(
    0,
    "Rank",
    range(1, len(importance_df) + 1)
)
```

```python
# Step 11: Display feature importance ranking

print("\nFeature Importance Ranking")
print("-" * 60)

print(importance_df.to_string(index=False))
```

```python
# Step 12: Export feature importance to CSV

importance_df.to_csv(
    "feature_importance.csv",
    index=False
)

print("\nFeature importance saved to feature_importance.csv")
```

```python
# Step 13: Display bootstrap configuration

print("\nRandom Forest Configuration")
print("-" * 60)

print("Number of trees:", model.n_estimators)
print("Bootstrap sampling:", model.bootstrap)
print("Features considered per split:", model.max_features)
print("Random state:", model.random_state)

print("\nExecution completed successfully.")
print("=" * 60)
```
