```python
import psutil
print("PID | Process Name | Status")
print("-" * 50)
count = 0
for process in psutil.process_iter(["pid", "name", "status"]):
    try:
        info = process.info
        print(info["pid"], "|", info["name"], " ", info["status"])
        count += 1
        if count == 15:
            break
    except (psutil.NoSuchProcess, psutil.AccessDenied):
        pass
```
