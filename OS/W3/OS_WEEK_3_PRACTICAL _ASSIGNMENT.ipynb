```python
import threading
import time

def deliver_order(order_number):
    thread_name = threading.current_thread().name

    print(f"{thread_name} is picking up Order-{order_number}")

    time.sleep(1)

    print(f"{thread_name} delivered Order-{order_number}")


threads = []

for i in range(1, 6):
    thread = threading.Thread(
        target=deliver_order,
        args=(i,),
        name=f"Agent-{i}"
    )
    threads.append(thread)
    thread.start()

for thread in threads:
    thread.join()

print("All 5 orders delivered!")
```
