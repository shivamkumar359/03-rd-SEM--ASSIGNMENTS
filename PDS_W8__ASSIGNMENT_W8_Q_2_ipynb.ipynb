```python
import pandas as pd
import plotly.express as px
```

```python
# Sample Heart Disease Dataset
data = {
    "Age": [43, 56, 68, 35],
    "Gender": ["Female", "Male", "Female", "Male"],
    "CP": [
        "Atypical Angina",
        "Typical Angina",
        "Asymptomatic",
        "Non-anginal"
    ],
    "Cholesterol": [341, 148, 411, 220],
    "Oldpeak": [3.0, 1.2, 3.6, 0.0]
}
```

```python
# Create a DataFrame
df = pd.DataFrame(data)

# Create an interactive scatter plot
fig = px.scatter(
    df,
    x="Age",
    y="Cholesterol",
    color="Gender",
    size="Oldpeak",
    hover_data=["CP", "Gender", "Age", "Cholesterol", "Oldpeak"],
    title="Heart Disease Analysis",
    labels={
        "Age": "Age (Years)",
        "Cholesterol": "Cholesterol Level",
        "Gender": "Gender",
        "CP": "Chest Pain Type",
        "Oldpeak": "Oldpeak"
    },
    size_max=40
)
```
