# **Question 1: IO Hardware & DMA**
A process needs to read a 200 MB file from an SSD into RAM.

(a) If polling is used, what does the CPU do during the entire 200 MB transfer?

(b) If interrupt-driven I/O with DMA is used, what does the CPU do during the transfer?

(c) Which DMA mode allows the CPU to continue accessing memory while DMA also uses the bus?

# **Answers 1: IO Hardware & DMA**

    (a) Polling: The CPU is stuck doing "busy-waiting." It sits in a continuous loop checking if the device is ready for the next byte, completely tying up the CPU so it cannot do any other work during the transfer.

    (b) Interrupt-driven I/O with DMA: The CPU simply gives the DMA controller the starting address and transfer size, then goes off to execute other processes. The CPU is only interrupted once when the DMA finishes the entire 200 MB transfer.

    (c) DMA Mode: Cycle Stealing mode. This allows the DMA to briefly take control of the bus for a single word transfer while the CPU continues its work.


# **Question 2: Buffering & Devices**
A student writes 10,000 bytes using f.write(byte) in a loop.

(a) Is this buffered or unbuffered I/O? Explain why.

(b) Is a hard disk a block device or a character device? What is its minimum transfer unit?

(c) What single change would make the write operation dramatically faster?

# **Answers 2: Buffering & Devices**

    (a) Buffering: It is generally buffered I/O. High-level languages gather these individual byte writes into a memory buffer and send them to the OS in larger chunks. If it were completely unbuffered, it would trigger 10,000 individual system calls, which would freeze the system.

    (b) Device Type: A hard disk is a block device. Its minimum transfer unit is a block or sector (typically 512 bytes or 4KB).

    (c) Speed Improvement: Write the data in larger chunks (like 4KB blocks) or write the entire 10,000 bytes array in a single operation instead of looping byte-by-byte.
