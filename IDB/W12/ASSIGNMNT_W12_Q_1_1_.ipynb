```python
# ============================================================
# TASK 1: MAPPING STRONG ENTITIES AND MULTIVALUED ATTRIBUTES
# IIT Jodhpur - Introduction to Databases
# ============================================================

import os
import zipfile
from google.colab import files

# Create output directory
OUTPUT_DIR = "/content/Task_1_Submission"
os.makedirs(OUTPUT_DIR, exist_ok=True)

# ------------------------------------------------------------
# FILE 1: SQL SCHEMA
# ------------------------------------------------------------

sql_content = """
-- TASK 1: Mapping Strong Entities and Multivalued Attributes

-- Create the strong entity table
CREATE TABLE DEPARTMENT (
    Dnumber INT PRIMARY KEY
);

-- Create a separate table for the multivalued Location attribute
CREATE TABLE DEPT_LOCATIONS (
    Dnumber INT NOT NULL,
    Dlocation VARCHAR(100) NOT NULL,

    PRIMARY KEY (Dnumber, Dlocation),

    FOREIGN KEY (Dnumber)
        REFERENCES DEPARTMENT(Dnumber)
        ON DELETE CASCADE
);

-- Example data insertion

INSERT INTO DEPARTMENT (Dnumber)
VALUES (1), (2), (3);

INSERT INTO DEPT_LOCATIONS (Dnumber, Dlocation)
VALUES
(1, 'Mumbai'),
(1, 'Delhi'),
(2, 'Bangalore'),
(2, 'Pune'),
(3, 'Hyderabad');

-- Display department records
SELECT * FROM DEPARTMENT;

-- Display department locations
SELECT * FROM DEPT_LOCATIONS;

-- Display departments with their locations
SELECT
    D.Dnumber,
    L.Dlocation
FROM DEPARTMENT D
JOIN DEPT_LOCATIONS L
ON D.Dnumber = L.Dnumber;
"""

sql_path = os.path.join(OUTPUT_DIR, "department_schema.sql")

with open(sql_path, "w") as f:
    f.write(sql_content)


# ------------------------------------------------------------
# FILE 2: TEXT-BASED SCHEMA
# ------------------------------------------------------------

schema_content = """
TASK 1: MAPPING STRONG ENTITIES AND MULTIVALUED ATTRIBUTES

1. DEPARTMENT TABLE

DEPARTMENT (
    Dnumber PRIMARY KEY
)

Primary Key:
Dnumber

Description:
DEPARTMENT is a strong entity. Dnumber uniquely identifies
each department.


2. DEPT_LOCATIONS TABLE

DEPT_LOCATIONS (
    Dnumber,
    Dlocation,
    PRIMARY KEY (Dnumber, Dlocation),
    FOREIGN KEY (Dnumber) REFERENCES DEPARTMENT(Dnumber)
)

Primary Key:
(Dnumber, Dlocation)

Foreign Key:
Dnumber references DEPARTMENT(Dnumber)


RELATIONSHIP:

One department can have multiple locations.
Each location record belongs to a department.

The separate DEPT_LOCATIONS relation handles the
multivalued Location attribute without storing multiple
values in a single column.
"""

txt_path = os.path.join(OUTPUT_DIR, "department_schema.txt")

with open(txt_path, "w") as f:
    f.write(schema_content)


# ------------------------------------------------------------
# FILE 3: PRIMARY KEY JUSTIFICATION REPORT
# Maximum 100 words
# ------------------------------------------------------------

report_content = """
PRIMARY KEY JUSTIFICATION REPORT

The DEPT_LOCATIONS table uses a composite primary key
consisting of Dnumber and Dlocation. Dnumber identifies
the department, while Dlocation identifies a particular
location associated with that department. Together, these
attributes uniquely identify each department-location
combination. This design allows one department to have
multiple locations while preventing duplicate location
entries for the same department. Additionally, Dnumber
acts as a foreign key referencing the DEPARTMENT table,
maintaining referential integrity between both relations.
"""

report_path = os.path.join(
    OUTPUT_DIR, "location_primary_key_report.md"
)

with open(report_path, "w") as f:
    f.write(report_content)


# ------------------------------------------------------------
# FILE 4: CREATE ZIP PACKAGE
# ------------------------------------------------------------

zip_path = "/content/Task_1_Database_Assignment.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zipf:

    zipf.write(
        sql_path,
        arcname="department_schema.sql"
    )

    zipf.write(
        txt_path,
        arcname="department_schema.txt"
    )

    zipf.write(
        report_path,
        arcname="location_primary_key_report.md"
    )


# ------------------------------------------------------------
# DISPLAY GENERATED FILES
# ------------------------------------------------------------

print("=" * 55)
print("TASK 1 - FILES GENERATED SUCCESSFULLY")
print("=" * 55)

for filename in os.listdir(OUTPUT_DIR):
    print("✓", filename)

print("\nZIP PACKAGE:", zip_path)

print("\nAll submission files are ready!")

# ------------------------------------------------------------
# AUTOMATICALLY DOWNLOAD ZIP
# ------------------------------------------------------------

files.download(zip_path)
```
