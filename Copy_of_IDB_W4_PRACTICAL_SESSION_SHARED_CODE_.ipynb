```python
# A list stores multiple values.
# Lists are ordered and mutable.
marks = [78, 85, 92, 66, 88]
```

```python
print("Original list:", marks)
```

```python
# Access
print("First mark:", marks[0])
```

```python
print("Last mark:", marks[-1])
```

```python
# Slicing
print("First 3 marks:", marks[:3])
```

```python
# Update
marks[1] = 90
print("After updating:", marks)
```

```python
# Add at the end
marks.append(95)
print("After append:", marks)
```

```python
# Insert at a position
marks.insert(2, 80)
print("After insert:", marks)
```

```python
# Remove a value
marks.remove(66)
print("After remove:", marks)
```

```python
# Remove last item
marks.pop()
print("After pop:", marks)
```

```python
# Sort
marks.sort()
print("Sorted marks:", marks)
```

```python
# Sort
marks.sort(reverse=False)
print("Sorted marks:", marks)
```

```python
# Length
print("Number of marks:", len(marks))
```

```python
# Total
print("Total:", sum(marks))
```

```python
print("Average:", sum(marks) / len(marks))
```

```python
# 4. SET
# ============================================================
print("\n========== 4. SET ==========")

# Set stores UNIQUE values.
# Sets are useful for mathematical operations.
```

```python
python_students = {
    "Rahul", "Priya", "Arun", "Sneha"
}
```

```python
sql_students = {
    "Priya", "Arun", "Kiran", "Meena"
}
```

```python
print("Python students:", python_students)
print("SQL students:", sql_students)
```

```python
# Add
python_students.add("Vijay")
print("After adding Vijay:", python_students)
```

```python
# Remove
python_students.remove("Vijay")
print("After removing Vijay:", python_students)
```

```python
# UNION
# Students in either group
all_students = python_students.union(sql_students)
print("All students:", all_students)
```

```python
# INTERSECTION
# Students common to both groups
common_students = python_students.intersection(sql_students)
print("Students learning both:", common_students)
```

```python
# DIFFERENCE
# Python students who are NOT in SQL
only_python = python_students.difference(sql_students)
print("Only Python:", only_python)
```

```python
# ============================================================
# 9. MEMORY EXPERIMENT
# ============================================================

print("\n========== MEMORY EXPERIMENT ==========")

# Different list sizes
sizes = [
    10,
    100,
    1000,
    5000,
    10000,
    50000,
    100000,
    500000
]
memory_usage = []
```

```python
# Create lists of different sizes
for size in sizes:

    my_list = list(range(size))

    # Measure list container memory
    memory = sys.getsizeof(my_list)

    memory_usage.append(memory)

    print(
        "Elements:",
        size,
        "| Memory:",
        memory,
        "bytes"
    )
```

```python
# ============================================================
# 10. MEMORY PLOT
# ============================================================

print("\n========== MEMORY PLOT ==========")

plt.figure(figsize=(10, 6))

plt.plot(
    sizes,
    memory_usage,
    marker="o"
)

plt.title(
    "List Length vs Memory Usage"
)

plt.xlabel(
    "Number of Elements"
)

plt.ylabel(
    "Memory Usage (Bytes)"
)

plt.grid(True)

plt.show()
```

```python
# 11. COMPARE MEMORY OF DATA STRUCTURES
# ============================================================
print("\n========== DATA STRUCTURE MEMORY ==========")
sample_list = list(range(1000))
sample_tuple = tuple(range(1000))
sample_set = set(range(1000))
sample_dictionary = {
    i: i for i in range(1000)
}

print(
    "List:",
    sys.getsizeof(sample_list),
    "bytes"
)

print(
    "Tuple:",
    sys.getsizeof(sample_tuple),
    "bytes"
)

print(
    "Set:",
    sys.getsizeof(sample_set),
    "bytes"
)

print(
    "Dictionary:",
    sys.getsizeof(sample_dictionary),
    "bytes"
)
```

```python
# ============================================================
# 6. COMPARING THE FOUR DATA STRUCTURES
# ============================================================
print("\n========== COMPARISON ==========")
my_list = [10, 20, 30]
my_tuple = (10, 20, 30)
my_set = {10, 20, 30}
my_dictionary = {
    "A": 10,
    "B": 20,
    "C": 30
}
print("LIST       :", my_list)
print("TUPLE      :", my_tuple)
print("SET        :", my_set)
print("DICTIONARY :", my_dictionary)
```

```python
# ============================================================
# 7. PRACTICAL EXAMPLE
# STUDENT ATTENDANCE
# ============================================================
print("\n========== STUDENT ATTENDANCE ==========")

attendance = {
    "Rahul": ["Present", "Present", "Absent"],
    "Priya": ["Present", "Present", "Present"],
    "Arun": ["Absent", "Present", "Present"]
}

for student, records in attendance.items():
    present = records.count("Present")
    total_classes = len(records)
    percentage = (
        present / total_classes
    ) * 100

    print(
        student,
        "->",
        percentage,
        "%"
    )
```
