```python
set1 = {"Math", 10, "Python", 20, 30}
set2 = {"Database", 20, "Python", 40, 50}

# Union of two sets
merged_set = set1.union(set2)

print("Set 1:", set1)
print("Set 2:", set2)
print("Merged Set:", merged_set)

# Length before adding duplicate
print("Length before adding duplicate:", len(merged_set))

# Adding a duplicate
merged_set.update({20})

print("Length after adding duplicate:", len(merged_set))

# Lambda function to cube only integer elements
cube = lambda x: x ** 3

cubed_values = [cube(x) for x in merged_set if type(x) == int]

print("Cubed integer values:", cubed_values)
```
