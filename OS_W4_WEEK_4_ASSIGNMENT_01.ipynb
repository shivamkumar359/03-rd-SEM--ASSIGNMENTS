```python
import threading

# Binary semaphore initialized to 0
semaphore = threading.Semaphore(0)


def P1():
    print("P1: Executing S1")
    print("P1: S1 completed")

    # Signal operation
    semaphore.release()


def P2():
    # Wait operation
    semaphore.acquire()

    print("P2: Executing S2")
    print("P2: S2 completed")


for run in range(1, 6):
    print(f"\n--- Run {run} ---")

    # Reset semaphore for each run
    semaphore = threading.Semaphore(0)

    p1 = threading.Thread(target=P1)
    p2 = threading.Thread(target=P2)

    p2.start()
    p1.start()

    p1.join()
    p2.join()
```
