```python
def process_student_marks(input_file, output_file):
    with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
        outfile.write("Name,Original Score,Processed Score\n")
        for line in infile:
            line = line.strip()
            if line:
                parts = line.split(',')
                name = parts[0]
                score = float(parts[1])
                processed_score = score * 1.05
                outfile.write(f"{name},{score},{processed_score:.2f}\n")

process_student_marks('marks.txt', 'processed_results.txt')
print("File processing complete. 'processed_results.txt' generated successfully.")
```
