```python
# ============================================================
# TASK 1: IMPLEMENT QUICK SORT AND MERGE SORT ALGORITHMS
# IIT JODHPUR BSc - DATA STRUCTURES / ALGORITHMS
# ============================================================

import os
import zipfile
from datetime import datetime
import matplotlib.pyplot as plt
```

```python
# ============================================================
# STEP 1: DEFINE QUICK SORT
# ============================================================

def quick_sort(arr, pivot_type="last", log=None):
    """
    Quick Sort using Divide and Conquer.

    pivot_type:
        last  = Last element as pivot
        first = First element as pivot
    """

    arr = arr.copy()

    def partition(low, high):
        if pivot_type == "last":
            pivot = arr[high]
            i = low - 1

            for j in range(low, high):
                if arr[j] <= pivot:
                    i += 1
                    arr[i], arr[j] = arr[j], arr[i]

            arr[i + 1], arr[high] = arr[high], arr[i + 1]

            return i + 1

        else:
            pivot = arr[low]

            i = low

            for j in range(low + 1, high + 1):
                if arr[j] <= pivot:
                    i += 1
                    arr[i], arr[j] = arr[j], arr[i]

            arr[low], arr[i] = arr[i], arr[low]

            return i

    def sort(low, high):
        if low < high:

            pivot_index = partition(low, high)

            if log is not None:
                log.append(
                    f"Pivot: {arr[pivot_index]}, "
                    f"Array: {arr.copy()}"
                )

            sort(low, pivot_index - 1)
            sort(pivot_index + 1, high)

    sort(0, len(arr) - 1)

    return arr
```

```python
# ============================================================
# STEP 2: DEFINE MERGE SORT
# ============================================================

def merge_sort(arr, log=None):
    """
    Merge Sort using Divide and Conquer.

    1. Divide the array into two halves.
    2. Recursively split until single elements remain.
    3. Merge the sorted halves.
    """

    arr = arr.copy()

    def sort(items):

        if len(items) <= 1:
            return items

        mid = len(items) // 2

        left = items[:mid]
        right = items[mid:]

        if log is not None:
            log.append(f"Splitting: {items}")
            log.append(f"  Left half: {left}")
            log.append(f"  Right half: {right}")

        left_sorted = sort(left)
        right_sorted = sort(right)

        merged = merge(left_sorted, right_sorted)

        if log is not None:
            log.append(f"Merging: {left_sorted} + {right_sorted}")
            log.append(f"  Result: {merged}")

        return merged

    def merge(left, right):

        result = []

        i = 0
        j = 0

        while i < len(left) and j < len(right):

            if left[i] <= right[j]:
                result.append(left[i])
                i += 1

            else:
                result.append(right[j])
                j += 1

        result.extend(left[i:])
        result.extend(right[j:])

        return result

    return sort(arr)
```

```python
# ============================================================
# STEP 3: INPUT DATASET
# ============================================================

original_array = [5, 3, 8, 1, 2]

print("=" * 60)
print("TASK 1: QUICK SORT AND MERGE SORT")
print("=" * 60)

print("\nOriginal Array:", original_array)
```

```python
# ============================================================
# STEP 4: EXECUTE QUICK SORT
# ============================================================

quick_log = []

quick_result = quick_sort(
    original_array,
    pivot_type="last",
    log=quick_log
)

print("\nQUICK SORT")
print("-" * 40)
print("Pivot Selection: Last Element")
print("Initial Array:", original_array)
print("Sorted Array:", quick_result)

print("\nQuick Sort Intermediate Steps:")

for step in quick_log:
    print(step)
```

```python
# ============================================================
# STEP 5: EXECUTE MERGE SORT
# ============================================================

merge_log = []

merge_result = merge_sort(
    original_array,
    log=merge_log
)

print("\nMERGE SORT")
print("-" * 40)
print("Initial Array:", original_array)
print("Sorted Array:", merge_result)

print("\nMerge Sort Intermediate Steps:")

for step in merge_log:
    print(step)
```

```python
# ============================================================
# STEP 6: COMPARE PIVOT SELECTION
# ============================================================

# Count partition comparisons for both pivot strategies.

def quick_sort_comparison_count(arr, pivot_type):

    arr = arr.copy()
    comparisons = 0

    def partition(low, high):

        nonlocal comparisons

        if pivot_type == "last":

            pivot = arr[high]
            i = low - 1

            for j in range(low, high):

                comparisons += 1

                if arr[j] <= pivot:
                    i += 1
                    arr[i], arr[j] = arr[j], arr[i]

            arr[i + 1], arr[high] = arr[high], arr[i + 1]

            return i + 1

        else:

            pivot = arr[low]
            i = low

            for j in range(low + 1, high + 1):

                comparisons += 1

                if arr[j] <= pivot:
                    i += 1
                    arr[i], arr[j] = arr[j], arr[i]

            arr[low], arr[i] = arr[i], arr[low]

            return i

    def sort(low, high):

        if low < high:

            pivot_index = partition(low, high)

            sort(low, pivot_index - 1)
            sort(pivot_index + 1, high)

    sort(0, len(arr) - 1)

    return comparisons


last_pivot_steps = quick_sort_comparison_count(
    original_array, "last"
)

first_pivot_steps = quick_sort_comparison_count(
    original_array, "first"
)

print("\nPIVOT EXPERIMENT")
print("-" * 40)

print("Last Element Pivot Comparisons:", last_pivot_steps)
print("First Element Pivot Comparisons:", first_pivot_steps)
```
