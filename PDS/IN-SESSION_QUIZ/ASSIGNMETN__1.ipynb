## Task : 1

1(a): Write a Python program to calculate the sum of numbers from 1 to n using a loop for n = 100, 1000, 10000, 100000.

```python
import time

# Different input sizes
input_sizes = [100, 1000, 10000, 100000]

# Run the program for each input size
for n in input_sizes:

    # Start the timer
    start_time = time.perf_counter()

    # Calculate the sum
    total = 0
    for i in range(1, n + 1):
        total += i

    # Stop the timer
    end_time = time.perf_counter()

    # Calculate execution time
    execution_time = end_time - start_time

    # Display the results
    print("Input size:", n)
    print("Sum:", total)
    print("Execution time:", execution_time, "seconds")
    print()
```

The following Python program uses a for loop to calculate the sum of numbers from 1 to n. It also measures the execution time for each input size.

```python
# 4(c)
# Given list
numbers = [42, 15, 8, 23, 4, 16]

# Bubble Sort
for i in range(len(numbers)):
    for j in range(0, len(numbers) - i - 1):

        # Compare adjacent elements
        if numbers[j] > numbers[j + 1]:

            # Swap the elements
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

# Display the sorted list
print("Sorted list:", numbers)
```

```python
# 6(c)


# Quick Sort function
def quick_sort(numbers):

    # If the list has 0 or 1 element, it is already sorted
    if len(numbers) <= 1:
        return numbers

    # Choose the first element as pivot
    pivot = numbers[0]

    smaller = []
    greater = []

    # Separate elements
    for value in numbers[1:]:
        if value <= pivot:
            smaller.append(value)
        else:
            greater.append(value)

    # Sort both groups and combine them
    return quick_sort(smaller) + [pivot] + quick_sort(greater)


# Given list
numbers = [25, 10, 30, 15, 5, 20, 35]

# Display the sorted list
print("Sorted list:", quick_sort(numbers))
```

## Task - 7

```python
import time
import random

# Bubble Sort function
def bubble_sort(numbers):
    for i in range(len(numbers)):
        for j in range(len(numbers) - i - 1):
            if numbers[j] > numbers[j + 1]:
                numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]


# Quick Sort function
def quick_sort(numbers):
    if len(numbers) <= 1:
        return numbers

    pivot = numbers[0]
    smaller = []
    greater = []

    for value in numbers[1:]:
        if value <= pivot:
            smaller.append(value)
        else:
            greater.append(value)

    return quick_sort(smaller) + [pivot] + quick_sort(greater)


# Input sizes
input_sizes = [100, 500, 1000]

for n in input_sizes:

    # Create a random list
    numbers = [random.randint(1, 10000) for _ in range(n)]

    # Copy the list for both algorithms
    bubble_numbers = numbers.copy()
    quick_numbers = numbers.copy()

    # Measure Bubble Sort time
    start_time = time.perf_counter()
    bubble_sort(bubble_numbers);
    bubble_time = time.perf_counter() - start_time

    # Measure Quick Sort time
    start_time = time.perf_counter()
    quick_sort(quick_numbers)
    quick_time = time.perf_counter() - start_time

    print("Input size:", n)
    print("Bubble Sort time:", bubble_time, "seconds")
    print("Quick Sort time:", quick_time, "seconds")
    print()
```

7(b)


Quick Sort is generally faster than Bubble Sort as the input size increases. Bubble Sort takes more time because it performs many comparisons and swaps. Quick Sort divides the list into smaller groups, which makes it faster for larger lists.

7(c)


Bubble Sort uses nested loops, so the number of operations increases quickly when the input size becomes larger. As the list grows, Bubble Sort needs many more comparisons and swaps. Therefore, its execution time increases more quickly than Quick Sort.

```python
# 7(d)


import matplotlib.pyplot as plt

# Example execution times
input_sizes = [100, 500, 1000]

bubble_times = [0.001, 0.025, 0.100]
quick_times = [0.0005, 0.002, 0.004]

# Plot the execution times
plt.plot(input_sizes, bubble_times, marker="o", label="Bubble Sort")
plt.plot(input_sizes, quick_times, marker="o", label="Quick Sort")

# Add labels
plt.xlabel("Input Size")
plt.ylabel("Execution Time (seconds)")
plt.title("Bubble Sort vs Quick Sort")

# Show legend and graph
plt.legend()
plt.show()
```

The graph shows that Quick Sort generally performs faster than Bubble Sort, especially when the input size becomes larger.
