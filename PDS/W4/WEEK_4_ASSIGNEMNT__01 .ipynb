```python
with open("marks.txt", "r") as file:
    marks = {}

    for line in file:
        line = line.strip()
        subject, mark = line.split(",")
        marks[subject] = int(mark)

print("Original Dictionary:")
print(marks)

print("\nDictionary Items:")
print(marks.items())

sorted_marks = dict(sorted(marks.items(), key=lambda item: item[1]))

print("\nSorted Dictionary:")
print(sorted_marks)

print("\nSorted Dictionary Items:")
print(sorted_marks.items())
```
