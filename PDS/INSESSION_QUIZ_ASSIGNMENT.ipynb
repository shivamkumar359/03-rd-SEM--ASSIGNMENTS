## task 1
Given input sizes

We need to calculate the sum of numbers from 1 to n for :    100, 1,000, 10,000, 100,000

```python
import time

# Different values of n that we want to test
input_sizes = [100, 1000, 10000, 100000]

for n in input_sizes:

    # Start the timer before running the loop
    start_time = time.time()

    total = 0

    # Add all numbers from 1 to n
    for i in range(1, n + 1):
        total += i

    # Stop the timer after the loop finishes
    end_time = time.time()

    execution_time = end_time - start_time

    print("Input size:", n)
    print("Sum:", total)
    print("Execution time:", execution_time, "seconds")
    print()
```
