```python
import numpy as np

# Create a 4x3 matrix containing values from 1 to 12
matrix = np.arange(1, 13).reshape(4, 3)

# Create the column index array
indices = np.array([0, 2, 0, 1])

# Select specific elements using advanced indexing
selected_elements = matrix[np.arange(4), indices]

# Display selected elements before addition
print("Selected Elements Before Addition:")
print(selected_elements)

# Add 10 to the selected elements in the original matrix
matrix[np.arange(4), indices] += 10

# Display the updated matrix
print("\nUpdated Matrix After Adding 10:")
print(matrix)
```
