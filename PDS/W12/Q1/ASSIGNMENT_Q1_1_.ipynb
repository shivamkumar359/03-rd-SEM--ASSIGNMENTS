```python
# ============================================================
# TASK 1: IMPLEMENT BFS GRAPH TRAVERSAL ALGORITHM
# IIT Jodhpur BSc - Introduction to Algorithms
# ============================================================

import os
import textwrap
import subprocess
import sys

# Create submission directory
os.makedirs("BFS_Task1", exist_ok=True)

# ============================================================
# PART 1: CREATE bfs_traversal.py
# ============================================================

bfs_code = r'''
from collections import deque

# ------------------------------------------------------------
# GRAPH REPRESENTATION
# ------------------------------------------------------------

graph = {
    1: [2, 3],
    2: [1, 4, 5],
    3: [1, 6],
    4: [2],
    5: [2],
    6: [3]
}

# ------------------------------------------------------------
# BFS ALGORITHM
# ------------------------------------------------------------

def bfs(graph, source):

    # Initialize all vertices
    color = {}
    distance = {}
    predecessor = {}

    for vertex in graph:
        color[vertex] = "WHITE"
        distance[vertex] = float('inf')
        predecessor[vertex] = None

    # Initialize source vertex
    color[source] = "GRAY"
    distance[source] = 0

    # FIFO queue
    queue = deque([source])

    traversal_order = []

    print("=" * 65)
    print("       BREADTH FIRST SEARCH (BFS) GRAPH TRAVERSAL")
    print("=" * 65)

    print("\\nInitial Source Vertex:", source)

    # BFS traversal
    while queue:

        # Remove first element (FIFO)
        u = queue.popleft()

        traversal_order.append(u)

        print(f"\\nProcessing Vertex: {u}")

        # Explore adjacent vertices
        for v in graph[u]:

            if color[v] == "WHITE":

                # Mark vertex as discovered
                color[v] = "GRAY"

                # Update shortest-path distance
                distance[v] = distance[u] + 1

                # Record predecessor
                predecessor[v] = u

                # Insert into FIFO queue
                queue.append(v)

        # Mark current vertex as completely processed
        color[u] = "BLACK"

        print("Queue:", list(queue))

    # --------------------------------------------------------
    # DISPLAY FINAL RESULTS
    # --------------------------------------------------------

    print("\\n" + "=" * 65)
    print("FINAL BFS TRAVERSAL RESULTS")
    print("=" * 65)

    print("\\nTraversal Order:")
    print(" -> ".join(map(str, traversal_order)))

    print("\\nVertex Details:")
    print("-" * 65)

    print(f"{'Vertex':<12}{'Color':<12}{'Distance':<15}{'Predecessor'}")
    print("-" * 65)

    for vertex in graph:

        pred = predecessor[vertex]

        if pred is None:
            pred = "NIL"

        print(
            f"{vertex:<12}"
            f"{color[vertex]:<12}"
            f"{distance[vertex]:<15}"
            f"{pred}"
        )

    print("-" * 65)

    return traversal_order, distance, predecessor, color


# ------------------------------------------------------------
# MAIN PROGRAM EXECUTION
# ------------------------------------------------------------

if __name__ == "__main__":

    source_vertex = 1

    bfs(
        graph,
        source_vertex
    )
'''

# Save Python file
py_file = "BFS_Task1/bfs_traversal.py"

with open(py_file, "w") as file:
    file.write(textwrap.dedent(bfs_code))

print("Created:", py_file)


# ============================================================
# PART 2: CREATE bfs_dry_run.txt
# ============================================================

dry_run = """
TASK 1: BFS GRAPH TRAVERSAL
MANUAL DRY RUN - FIRST THREE ITERATIONS

Graph Adjacency List:

1 -> 2, 3
2 -> 1, 4, 5
3 -> 1, 6
4 -> 2
5 -> 2
6 -> 3

Source Vertex: 1

Initial Conditions:
-------------------
All vertices are WHITE.
Distance of every vertex = Infinity.
Predecessor of every vertex = NIL.

Source vertex 1:
Color = GRAY
Distance = 0
Predecessor = NIL

Initial Queue:
[1]


ITERATION 1
===========

Dequeue vertex 1.

Current Vertex: 1
Neighbors: 2, 3

Vertex 2:
- Initially WHITE.
- Change color to GRAY.
- Distance = distance[1] + 1 = 1.
- Predecessor = 1.
- Enqueue vertex 2.

Vertex 3:
- Initially WHITE.
- Change color to GRAY.
- Distance = distance[1] + 1 = 1.
- Predecessor = 1.
- Enqueue vertex 3.

Mark vertex 1 as BLACK.

Queue after Iteration 1:
[2, 3]

Discovered vertices: 1, 2, 3


ITERATION 2
===========

Dequeue vertex 2.

Current Vertex: 2
Neighbors: 1, 4, 5

Vertex 1:
- Already BLACK.
- Ignore.

Vertex 4:
- Initially WHITE.
- Change color to GRAY.
- Distance = distance[2] + 1 = 2.
- Predecessor = 2.
- Enqueue vertex 4.

Vertex 5:
- Initially WHITE.
- Change color to GRAY.
- Distance = distance[2] + 1 = 2.
- Predecessor = 2.
- Enqueue vertex 5.

Mark vertex 2 as BLACK.

Queue after Iteration 2:
[3, 4, 5]

Discovered vertices: 1, 2, 3, 4, 5


ITERATION 3
===========

Dequeue vertex 3.

Current Vertex: 3
Neighbors: 1, 6

Vertex 1:
- Already BLACK.
- Ignore.

Vertex 6:
- Initially WHITE.
- Change color to GRAY.
- Distance = distance[3] + 1 = 2.
- Predecessor = 3.
- Enqueue vertex 6.

Mark vertex 3 as BLACK.

Queue after Iteration 3:
[4, 5, 6]

Discovered vertices: 1, 2, 3, 4, 5, 6


FINAL EXPECTED RESULTS
======================

BFS Traversal Order:
1 -> 2 -> 3 -> 4 -> 5 -> 6

Vertex    Distance    Predecessor
1         0           NIL
2         1           1
3         1           1
4         2           2
5         2           2
6         2           3


CONCLUSION
==========

The BFS algorithm successfully traverses the graph using
a First-In, First-Out (FIFO) queue.

All vertices are visited in breadth-first order.

The algorithm correctly implements:
1. White, Gray and Black vertex states.
2. Shortest-path distance calculation.
3. Predecessor tracking.
4. FIFO queue operations.

The algorithm runs in O(V + E) time using an adjacency list,
where V represents vertices and E represents edges.
"""

txt_file = "BFS_Task1/bfs_dry_run.txt"

with open(txt_file, "w") as file:
    file.write(textwrap.dedent(dry_run))

print("Created:", txt_file)


# ============================================================
# PART 3: EXECUTE BFS PYTHON PROGRAM
# ============================================================

print("\n\nEXECUTING BFS PROGRAM...\n")

result = subprocess.run(
    [sys.executable, py_file],
    capture_output=True,
    text=True
)

print(result.stdout)

if result.returncode != 0:
    print("Execution Error:")
    print(result.stderr)
else:
    print("BFS PROGRAM EXECUTED SUCCESSFULLY!")


# ============================================================
# PART 4: DISPLAY DRY RUN FILE
# ============================================================

print("\n\n" + "=" * 65)
print("CONTENTS OF bfs_dry_run.txt")
print("=" * 65)

with open(txt_file, "r") as file:
    print(file.read())


# ============================================================
# PART 5: DOWNLOAD SUBMISSION FILES
# ============================================================

print("\n" + "=" * 65)
print("SUBMISSION FILES GENERATED")
print("=" * 65)

print("\n1.", py_file)
print("2.", txt_file)

from google.colab import files

print("\nDownload bfs_traversal.py")
files.download(py_file)

print("\nDownload bfs_dry_run.txt")
files.download(txt_file)

print("\nTASK 1 FILE GENERATION COMPLETED!")
```
