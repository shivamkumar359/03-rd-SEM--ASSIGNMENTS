# Q.1 Description:
Create a Python script that reads student subject and marks data from a text file, stores it in a dictionary, and performs basic analysis including sorting by key and calculating frequencies.

Expected Deliverables:

    Python script named "marks_processor.py" containing the dictionary implementation and sorting logic.

    A text file named "marks.txt" with at least five lines of comma-separated subject and mark pairs.

    Screenshot of the console output showing the dictionary contents before and after using the popitem() method.

    Screenshot of the console output displaying the sorted list of subject keys using the sorted() function.

Success Criteria:

    Script successfully converts raw text file lines into a Python dictionary with unique subject keys.

    Implementation correctly uses the .get() method to retrieve marks and handles non-existent keys by returning None.

    The output demonstrates that the popitem() function removes the last inserted key-value pair by default.


```python
# marks_processor.py

# Read subject and marks from marks.txt
marks = {}

with open("marks.txt", "r") as file:
    for line in file:
        subject, mark = line.strip().split(",")
        marks[subject] = int(mark)

# Display original dictionary
print("Dictionary before popitem():")
print(marks)

# Demonstrate .get() method
print("\nMath marks:", marks.get("Math"))
print("Non-existent subject marks:", marks.get("History"))

# Demonstrate popitem()
removed_item = marks.popitem()

print("\nRemoved item using popitem():")
print(removed_item)

print("\nDictionary after popitem():")
print(marks)

# Sort subject keys
sorted_subjects = sorted(marks.keys())

print("\nSorted list of subject keys:")
print(sorted_subjects)
```
