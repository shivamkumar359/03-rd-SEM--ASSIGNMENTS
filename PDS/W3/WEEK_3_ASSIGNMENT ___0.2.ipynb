# Q.2 Description:
Develop a set of functions to perform character frequency counting and string slicing to verify specific prefixes and suffixes within a given text string.

Expected Deliverables:

    Python script named "string_analyzer.py" containing a custom frequency counter and slicing logic.

    Screenshot showing the frequency count of a specific character (e.g., 'L' in "Hello World").

    Screenshot showing the boolean results of the .startswith() and .endswith() functions on a test string.

    A short text file "slicing_results.txt" showing the output of extracting the first four characters of a string.

Success Criteria:

    The frequency function accurately counts occurrences of a specified character within a provided string.

    Slicing logic correctly identifies and returns substrings based on start and end position arguments.

    The script correctly utilizes built-in string methods to validate if a string begins or ends with a specific character.


```python
# string_analyzer.py

# Function to count frequency of a character
def character_frequency(text, character):
    count = 0

    for ch in text:
        if ch == character:
            count += 1

    return count


# Function to perform string slicing
def slice_string(text, start, end):
    return text[start:end]


# Test string
text = "Hello World"

# Character frequency
character = "l"
frequency = character_frequency(text, character)

print("Text:", text)
print("Frequency of", character, ":", frequency)

# String slicing
first_four = slice_string(text, 0, 4)

print("\nFirst four characters:", first_four)

# startswith() and endswith()
print("\nStarts with 'H':", text.startswith("H"))
print("Ends with 'd':", text.endswith("d"))
```
