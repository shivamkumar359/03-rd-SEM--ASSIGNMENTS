```python
def calculate_factorial(n):
    if n < 0:
        return "Invalid: Negative number"
    elif n == 0 or n == 1:
        return 1
    else:
        return n * calculate_factorial(n - 1)

def process_input(n):
    result = calculate_factorial(n)
    # Storing output in a tuple: (integer input, string identifier, result value as float or string)
    if isinstance(result, int):
        output_tuple = (n, f"Factorial of {n}", float(result))
    else:
        output_tuple = (n, "Error State", result)
    return output_tuple

# Test cases for inputs 6, 0, and -1
inputs = [6, 0, -1]
for val in inputs:
    print(process_input(val))
```
