```python
clean_text = """============================================================
DATABASE SCHEMA DEFINITIONS: UNIVERSITY DEPARTMENT
============================================================

1. TABLE: student
------------------------------------------------------------
Attributes & Properties:
- student_id   : VARCHAR(20)  | NOT NULL | PRIMARY KEY
- student_login: VARCHAR(50)  | NOT NULL | UNIQUE
- first_name   : VARCHAR(50)  | NOT NULL
- last_name    : VARCHAR(50)  | NOT NULL
- email        : VARCHAR(100) | NOT NULL | UNIQUE
- gpa          : DECIMAL(3,2) | NULL

Primary Key:
- student_id


2. TABLE: instructor
------------------------------------------------------------
Attributes & Properties:
- instructor_id: VARCHAR(20)  | NOT NULL | PRIMARY KEY
- first_name   : VARCHAR(50)  | NOT NULL
- last_name    : VARCHAR(50)  | NOT NULL
- email        : VARCHAR(100) | NOT NULL | UNIQUE
- office_number: VARCHAR(20)  | NULL
- salary       : DECIMAL(10,2)| NULL

Primary Key:
- instructor_id
============================================================
"""

with open("schema_definitions.txt", "w") as f:
    f.write(clean_text)

print("schema_definitions.txt cleaned and saved.")
```

```python
import sqlite3
import pandas as pd

# 1. Connect to an in-memory SQLite database
conn = sqlite3.connect(":memory:")

# 2. Create the student table
conn.execute("""
CREATE TABLE student (
    student_id VARCHAR(20) PRIMARY KEY,
    student_login VARCHAR(50) NOT NULL UNIQUE,
    first_name VARCHAR(50) NOT NULL,
    last_name VARCHAR(50) NOT NULL,
    email VARCHAR(100) NOT NULL UNIQUE,
    gpa DECIMAL(3,2)
);
""")

# 3. Retrieve and format table structure metadata
df = pd.read_sql_query("PRAGMA table_info(student);", conn)
df = df[["cid", "name", "type", "notnull", "dflt_value", "pk"]]
df
```
